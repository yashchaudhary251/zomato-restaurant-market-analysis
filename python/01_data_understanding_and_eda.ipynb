### Restaurant Market & Customer Experience Analysis

```python
# Which cuisines are most popular in the dataset?
most_popular_cuisine = df['Cuisines'].value_counts().sort_values(ascending = False)
most_popular_cuisine.head(15)
```

```python
# Which cities have the highest average restaurant rating?

highest_avg_rating_restaurant_cities = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending = False)
highest_avg_rating_restaurant_cities.head(15)
```

```python
# Which cities have a large number of restaurants while also maintaining a good average rating?


city_analysis = df.groupby('City').agg(
    Restaurant_Count=('Restaurant ID', 'count'),
    Average_Rating=('Aggregate rating', 'mean')
)

city_analysis.sort_values(by='Restaurant_Count' , ascending = False).head(15)
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns 

top_10_cities = city_analysis.sort_values(by='Average_Rating',ascending = False).head(10)

sns.barplot(data = top_10_cities , x= 'City',y = 'Average_Rating')
plt.title("Top 10 Cities by Average Rating")
plt.xlabel("Cities")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
# Which cuisines have a high average rating and a sufficient number of restaurants?

highest_rated_cuisine = df.groupby('Cuisines').agg(
    restaurant_count = ('Restaurant ID','count'),
    average_rating = ('Aggregate rating', 'mean')
)

at_least_20_restaurant = highest_rated_cuisine[highest_rated_cuisine['restaurant_count'] >= 20]

at_least_20_restaurant.sort_values(by='average_rating',ascending = False).head(15)
```

```python
# What are the top-rated cuisines among cuisines with at least 20 restaurants?

top_rated_cuisines = at_least_20_restaurant.sort_values(by='average_rating',ascending = False).head(10)

sns.barplot(data = top_rated_cuisines , x= 'Cuisines',y = 'average_rating')
plt.title("Top 10 cuisines by average rating, considering only cuisines with at least 20 restaurants.")
plt.xlabel("Cuisines Name")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
# How does restaurant price range relate to average restaurant rating?

price_range_relation_to_ARR = df.groupby('Price range')['Aggregate rating'].mean()
price_range_relation_to_ARR 
```

```python
# Does online delivery availability vary by city?

availability_of_OD = pd.crosstab(
    df['City'],
    df['Has Online delivery'],
    normalize='index'
) * 100

availability_of_OD = availability_of_OD.round(1)

availability_of_OD.head(15)
```

```python
city_restaurant_count = df.groupby('City')['Restaurant ID'].count()
cities_10_plus = city_restaurant_count[city_restaurant_count >= 10].index
availability_of_OD_10_plus = availability_of_OD.loc[
    availability_of_OD.index.isin(cities_10_plus)
]
availability_of_OD_10_plus.sort_values(by='Yes',ascending =False).head(15)
```

```python
# Among cities with at least 10 restaurants, which cities have the highest percentage of restaurants offering online delivery?


top_cities_OD = availability_of_OD_10_plus.sort_values(
    by='Yes',
    ascending=False
).head(10)

sns.barplot(
    data=top_cities_OD.reset_index(),
    x='Yes',
    y='City'
)

plt.title("Top 10 Cities by Online Delivery Availability")
plt.xlabel("Online Delivery Available (%)")
plt.ylabel("City")
plt.show()
```

```python
# Does table booking availability vary across different cities?

availability_of_TB = pd.crosstab(
    df['City'],
    df['Has Table booking'],
    normalize='index'
) * 100

availability_of_TB = availability_of_TB.round(1)

city_restaurant_count = df.groupby('City')['Restaurant ID'].count()
cities_10_plus = city_restaurant_count[city_restaurant_count >= 10].index
availability_of_TB_10_plus = availability_of_TB.loc[
    availability_of_TB.index.isin(cities_10_plus)
]
availability_of_TB_10_plus.sort_values(by='Yes',ascending =False).head(15)

availability_of_TB.head(15)
```

```python
# Among cities with at least 10 restaurants, which cities have the highest percentage of restaurants offering Table bookingg?


top_cities_TB = availability_of_TB_10_plus.sort_values(
    by='Yes',
    ascending=False
).head(10)

sns.barplot(
    data=top_cities_TB.reset_index(),
    x='Yes',
    y='City'
)

plt.title("Top 10 Cities by Table Booking Availability")
plt.xlabel("Table Booking Available (%)")
plt.ylabel("City")
plt.show()
```
